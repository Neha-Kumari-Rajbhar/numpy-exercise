```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame({
"City":["Delhi","Delhi","Delhi","Mumbai","Mumbai","Mumbai"],
"Product":["Shoes","Shoes","Watch","Shoes","Watch","Watch"],
"Sales":[200,150,300,400,250,300]
})
```

```python
df.pivot_table(values="Sales", index="City", columns="Product", aggfunc="sum",fill_value=0)
```

```python
data = pd.DataFrame({
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
})
data
```

```python
data["Month"]=data['Date'].dt.month_name()
data
```

```python
data['Month']=data['Date'].dt.month
data
```

```python
#data["Quater"]=data["Date"].dt.quarter
data["Quater"]='Q'+ data['Date'].dt.quarter.astype(str)
data
```

```python
#Sale of products in each region
pd.pivot_table(data,values='Sales',index="Region",columns='Product',aggfunc='mean')
```

```python
data
```

```python
pd.pivot_table(data,values=["Sales","Units"],index="Product",columns="Region",aggfunc="median")
```

```python
pd.crosstab(data['Product'],data["Region"])
```

```python
data2=pd.crosstab(data["Region"],data["Product"],values=data["Sales"],aggfunc="sum")
data2
```
