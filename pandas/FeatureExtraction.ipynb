```python
#make a new column for episode count
df['Title']
```

```python
def extract_eps(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True: 
            data = data + i 
        if i == '(':
            check = True 
```

```python
df["Episodes"]=df['Title'].apply(extract_eps)
df.head()
```

```python
df["Episodes"]=df["Episodes"].str.replace("eps","")
df.head()
```

```python
type(df.loc[0]['Episodes'])
```

```python
df['Episodes']=df['Episodes'].astype(int)
type(df.loc[0]['Episodes'])
df.head()
```

```python
#make a new column for time stamp
df['Title'][0]
```

```python
# def extraction_time(txt):
#     check = False
#     data = ""
#     for i in range(len(txt)):
#         if txt[i] == ')':
#             for j in range(i+1,i+20):
#                 data += txt[j]

#             return data

def time_count(text):
    ans=""
    for i in range(0,len(text)):
        if text[i]==")":
            for j in range(i+1,i+20):
                ans=ans+text[j]
                
            return(ans)
```

```python
df['time-stamp']=df["Title"].apply(time_count)
df.head()
```

```python
df[['start_date','end_date']]=df['time-stamp'].str.split("-",expand=True) #for splitting values based on -
df.head()
```

```python
df['start_date'] = df['start_date'].str.strip() #strip----> for removing spaces
df['end_date'] = df['end_date'].str.strip()

df['start_date'] = pd.to_datetime(df['start_date'], format='%b %Y')
df['end_date'] = pd.to_datetime(df['end_date'], format='%b %Y')
df.head()
```

```python
df['total_months']=(df['end_date'].dt.year - df['start_date'].dt.year)*12 + (df['end_date'].dt.month - df['start_date'].dt.month)+1
df
```
